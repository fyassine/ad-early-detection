# Four-Tier Evaluation Protocol — Vector Plot Assets Generator

Follows  strictly:
- Visual Reference: 
- Typography: Monospace 
- Palettes: Two-hue Teal / Purple ( / ) and Okabe-Ito Vermillion ()
- No red-green pairings
- Inward ticks, top/right spines removed
- TrueType PDF export ()
- Anti-occlusion bboxes and point-based offsets


```python
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import numpy as np

OUTPUT_DIR = Path("../plots_tkiz/protocol_funnel").resolve()
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
MM = 1 / 25.4

COLOR_TEAL_FILL = "#2ba099"
COLOR_TEAL_EDGE = "#14605b"
COLOR_PURPLE_FILL = "#873397"
COLOR_PURPLE_EDGE = "#50165b"
COLOR_GRAY_FILL = "#DAD7CB"
COLOR_GRAY_EDGE = "#808080"
COLOR_VERMILLION = "#D55E00"
COLOR_DARK = "#222222"
TEXT_BBOX = dict(boxstyle="round,pad=0.20", facecolor="white", edgecolor="none", alpha=0.88)

plt.rcParams.update({
    "font.family": "Courier New",
    "font.size": 7.5,
    "axes.labelsize": 8.0,
    "xtick.labelsize": 7.0,
    "ytick.labelsize": 7.5,
    "legend.fontsize": 7.0,
    "axes.linewidth": 0.8,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 3.0,
    "ytick.major.size": 3.0,
    "legend.frameon": False,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "svg.fonttype": "none",
    "text.color": COLOR_DARK,
})
print("✓ Style parameters loaded successfully.")
```

### 1. Tier 1 Floor Gates Plot


```python
fig, ax = plt.subplots(figsize=(40 * MM, 23 * MM))
y_pos = [0, 1, 2]
vals = [0.530, 0.492, 0.749]
labels = ["Demo", "Static", "S1"]
colors = [COLOR_GRAY_FILL, COLOR_GRAY_FILL, COLOR_TEAL_FILL]
edgecolors = [COLOR_GRAY_EDGE, COLOR_GRAY_EDGE, COLOR_TEAL_EDGE]
height = 0.55

for y, v, c, ec in zip(y_pos, vals, colors, edgecolors):
    ax.barh(y, v, height=height, color=c, edgecolor=ec, linewidth=0.8, zorder=3)

ax.annotate("0.53", xy=(0.53, 0), xytext=(4, 0), textcoords="offset points", va="center", ha="left", fontsize=6.5, color=COLOR_DARK, zorder=10)
ax.annotate("0.49", xy=(0.49, 1), xytext=(4, 0), textcoords="offset points", va="center", ha="left", fontsize=6.5, color=COLOR_DARK, zorder=10)
ax.annotate("0.75", xy=(0.749, 2), xytext=(-5, 0), textcoords="offset points", va="center", ha="right", fontsize=6.8, fontweight="bold", color="white", zorder=10)
ax.axvline(0.530, color=COLOR_VERMILLION, linestyle="--", linewidth=0.85, zorder=2)
ax.annotate("hurdle (0.53)", xy=(0.530, 2.50), xytext=(3, 0), textcoords="offset points", fontsize=6.2, color=COLOR_VERMILLION, fontweight="bold", va="center", bbox=TEXT_BBOX, zorder=10)

ax.set_yticks(y_pos)
ax.set_yticklabels(labels)
ax.set_xlim(0, 0.95)
ax.set_ylim(-0.50, 2.80)
ax.set_xlabel("OOF AUC", labelpad=1)
ax.set_xticks([0.0, 0.5, 0.75])
ax.set_xticklabels(["0", ".50", ".75"])
ax.spines[["top", "right"]].set_visible(False)
fig.tight_layout(pad=0.25)
fig.savefig(OUTPUT_DIR / "tier1_floor_gates.pdf", format="pdf", bbox_inches="tight", pad_inches=0.02)
plt.show()
```

### 2. Tier 2 Selection Rule Plot


```python
fig, ax = plt.subplots(figsize=(40 * MM, 23 * MM))
seed_deltas = [0.031, 0.008, 0.024, -0.005]
seed_y = [3, 2, 1, 0]
ax.axvline(0, color=COLOR_GRAY_EDGE, linestyle=":", linewidth=0.65, zorder=1)
se = 0.0074
mean_val = 0.0145
ax.axvspan(se, 0.045, color=COLOR_TEAL_FILL, alpha=0.10, zorder=0)
ax.axvline(se, color=COLOR_VERMILLION, linestyle="--", linewidth=0.85, zorder=2)
ax.annotate("1-SE threshold", xy=(se, 3.85), xytext=(2, 0), textcoords="offset points", fontsize=6.0, color=COLOR_VERMILLION, fontweight="bold", va="center", bbox=TEXT_BBOX, zorder=10)
for y, d in zip(seed_y, seed_deltas):
    ax.scatter(d, y, s=18, color=COLOR_TEAL_FILL, edgecolor=COLOR_TEAL_EDGE, linewidth=0.7, zorder=4, alpha=0.9)
y_mean = 4.20
ax.errorbar(mean_val, y_mean, xerr=se, fmt="D", markersize=4.2, color=COLOR_PURPLE_FILL, ecolor=COLOR_PURPLE_FILL, markeredgecolor=COLOR_PURPLE_EDGE, elinewidth=1.2, capsize=2.5, capthick=0.8, zorder=5)
ax.set_yticks([0, 1, 2, 3, y_mean])
ax.set_yticklabels(["s4", "s3", "s2", "s1", "Mean"])
ax.set_xlim(-0.025, 0.045)
ax.set_ylim(-0.45, 4.75)
ax.set_xlabel(r"$\Delta\mathrm{AUC}$", labelpad=1)
ax.set_xticks([-0.02, 0.0, 0.02, 0.04])
ax.set_xticklabels(["-.02", "0", "+.02", "+.04"])
ax.spines[["top", "right"]].set_visible(False)
fig.tight_layout(pad=0.25)
fig.savefig(OUTPUT_DIR / "tier2_selection_rule.pdf", format="pdf", bbox_inches="tight", pad_inches=0.02)
plt.show()
```
