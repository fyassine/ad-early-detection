## Cohort summary statistics

For each of (1) the whole DELCODE cohort, (2) the pretrain split pool, and (3) the downstream split pool:

- number of patients and number of scans
- scans per patient (distribution)
- number of patients per diagnosis, and number of scans per diagnosis (whole cohort only — scan-level)
- number of patients and scans per individual split (train/val/test) for the two split pools

Sections A–C print these as text. Sections D–E re-render the same numbers as side-by-side tables and plots for easier comparison across cohorts.

These are descriptive only — no assertions — and exist to make cohort composition visible alongside the hygiene checks above.

```python
def summarize_scan_level_cohort(df, id_col='Pseudonym', diagnosis_col='diagnosis', label=''):
    """Summarize a scan-level table (one row per scan, e.g. cohorts.csv).

    Prints the text report and returns the underlying series/frames so they can be
    reused for the side-by-side tables and plots below.
    """
    scans_per_patient = df.groupby(id_col).size()
    diagnosis_patients = df.drop_duplicates([id_col, diagnosis_col])[diagnosis_col].value_counts()
    diagnosis_scans = df[diagnosis_col].value_counts()

    print(f'--- {label} ---')
    print(f'Patients: {df[id_col].nunique()}')
    print(f'Scans:    {len(df)}')

    print('\nScans per patient — describe():')
    print(scans_per_patient.describe())
    print('\nScans per patient — distribution (n_scans -> n_patients):')
    print(scans_per_patient.value_counts().sort_index())

    print(f'\nPatients per diagnosis (unique {id_col} x {diagnosis_col} pairs — a patient who '
          'converted/progressed across visits is counted under each diagnosis they carried):')
    print(diagnosis_patients)

    print(f'\nScans per diagnosis (raw row counts by {diagnosis_col}):')
    print(diagnosis_scans)
    print()

    return {
        'n_patients': df[id_col].nunique(),
        'n_scans': len(df),
        'scans_per_patient': scans_per_patient,
        'diagnosis_patients': diagnosis_patients,
        'diagnosis_scans': diagnosis_scans,
    }


def summarize_split_pool(split_csvs, label=''):
    """Summarize a {train,val,test} -> csv mapping where each row is already one patient with an n_scans column.

    Prints the text report and returns the underlying frames/series so they can be
    reused for the side-by-side tables and plots below.
    """
    dfs = {name: pd.read_csv(path) for name, path in split_csvs.items()}
    pool = pd.concat(dfs.values(), ignore_index=True)

    per_split = pd.DataFrame({
        name: {'patients': len(d), 'scans': int(d['n_scans'].sum())}
        for name, d in dfs.items()
    }).T
    per_split.loc['total'] = per_split.sum()

    diagnosis_per_split = pd.DataFrame({
        name: d['diagnosis'].value_counts() for name, d in dfs.items()
    }).fillna(0).astype(int).T
    diagnosis_per_split.loc['total'] = diagnosis_per_split.sum()

    print(f'--- {label} ---')
    print('Patients and scans per split:')
    print(per_split)

    print('\nScans per patient — describe():')
    print(pool['n_scans'].describe())
    print('\nScans per patient — distribution (n_scans -> n_patients):')
    print(pool['n_scans'].value_counts().sort_index())

    print('\nPatients per diagnosis:')
    print(pool['diagnosis'].value_counts())

    print('\nPatients per diagnosis, per split:')
    print(diagnosis_per_split)
    print()

    return {
        'n_patients': pool['Pseudonym'].nunique(),
        'n_scans': int(pool['n_scans'].sum()),
        'pool': pool,
        'per_split': per_split,
        'diagnosis_per_split': diagnosis_per_split,
        'scans_per_patient': pool['n_scans'],
        'diagnosis_patients': pool['diagnosis'].value_counts(),
    }
```

### D. Side-by-side comparison tables

Same numbers as the text reports above (A–C), laid out as tables so the three cohorts can be compared at a glance.

```python
from IPython.display import display, display_html


def display_side_by_side(dfs, titles=()):
    """Render several small DataFrames as HTML tables laid out in a row."""
    html = '<div style="display:flex; gap:40px; align-items:flex-start;">'
    for df, title in zip(dfs, titles):
        html += '<div>'
        if title:
            html += f'<h4>{title}</h4>'
        html += df.to_html()
        html += '</div>'
    html += '</div>'
    display_html(html, raw=True)


# --- Table 1: overview (patients / scans per cohort) ---
overview = pd.DataFrame({
    'Whole DELCODE': {'patients': whole_stats['n_patients'], 'scans': whole_stats['n_scans']},
    'Pretrain pool': {'patients': pretrain_stats['n_patients'], 'scans': pretrain_stats['n_scans']},
    'Downstream pool': (
        {'patients': downstream_stats['n_patients'], 'scans': downstream_stats['n_scans']}
        if downstream_stats is not None else {'patients': 0, 'scans': 0}
    ),
}).T
print('Overview — patients & scans per cohort:')
display(overview)

# --- Table 2: patients/scans per diagnosis, side by side across cohorts ---
diag_cols = {'Whole DELCODE (patients)': whole_stats['diagnosis_patients'],
             'Whole DELCODE (scans)': whole_stats['diagnosis_scans'],
             'Pretrain pool (patients)': pretrain_stats['diagnosis_patients'],
             'Pretrain pool (scans)': pretrain_stats['pool'].groupby('diagnosis')['n_scans'].sum()}
if downstream_stats is not None:
    diag_cols['Downstream pool (patients)'] = downstream_stats['diagnosis_patients']
    diag_cols['Downstream pool (scans)'] = downstream_stats['pool'].groupby('diagnosis')['n_scans'].sum()
diagnosis_table = pd.DataFrame(diag_cols).fillna(0).astype(int)
print('\nPatients/scans per diagnosis, by cohort:')
display(diagnosis_table)

# --- Table 3: per-split patients & scans, pretrain vs downstream, side by side ---
tables = [pretrain_stats['per_split']]
titles = ['Pretrain split pool']
if downstream_stats is not None:
    tables.append(downstream_stats['per_split'])
    titles.append('Downstream split pool')
display_side_by_side(tables, titles=titles)
```

### E. Plots

Visual versions of tables D2 (diagnosis composition), the scans-per-patient distributions, and table D3 (per-split sizes).

```python
import matplotlib.pyplot as plt
import numpy as np

DIAG_ORDER = ['ad', 'healthy', 'mci', 'converter', 'other/excluded']
SCAN_ON_DISK_NOTE = ('Note: patients included must have the physical scan (.npz connectivity '
                      'matrix) on disk — cohorts_with_scans_on_disk.csv / the split CSVs already filter to this set.')


def _note_center_x(fig, axes):
    """X-position (figure fraction) centered on the plotted axes, not the bare figure canvas.

    fig.text(0.5, ...) centers on the whole figure, which is off-center relative to the
    actual plot area once y-axis labels/ticks eat into the left margin. Call this after
    fig.tight_layout() so ax.get_position() reflects the final layout.
    """
    ax_list = np.atleast_1d(axes).ravel().tolist()
    fig.canvas.draw()
    x0 = min(ax.get_position().x0 for ax in ax_list)
    x1 = max(ax.get_position().x1 for ax in ax_list)
    return (x0 + x1) / 2


def _plot_diag_stack(ax, crosstab, title, xlabel):
    """Stacked bar of a (row -> diagnosis counts) crosstab, with the row total labeled on top."""
    crosstab = crosstab.reindex(columns=DIAG_ORDER, fill_value=0)
    crosstab.plot(kind='bar', stacked=True, ax=ax, legend=False,
                  color=[plt.cm.tab10(i) for i in range(len(DIAG_ORDER))])
    totals = crosstab.sum(axis=1)
    for i, total in enumerate(totals):
        ax.text(i, total, str(int(total)), ha='center', va='bottom', fontsize=8)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel('# patients')


# --- Plot 1: patients per diagnosis, grouped bar across cohorts ---
patient_cols = [c for c in diagnosis_table.columns if c.endswith('(patients)')]
plot_df = diagnosis_table[patient_cols].sort_index()

fig, ax = plt.subplots(figsize=(8, 4.5))
plot_df.plot(kind='bar', ax=ax)
for container in ax.containers:
    ax.bar_label(container, fontsize=8)
ax.set_title('Patients per diagnosis, by cohort')
ax.set_xlabel('diagnosis')
ax.set_ylabel('# patients')
ax.legend(title=None, fontsize='small')

whole_ad = diagnosis_table['Whole DELCODE (patients)'].get('ad', 0)
pretrain_ad = diagnosis_table['Pretrain pool (patients)'].get('ad', 0)
note_y = -0.05
fig.tight_layout(rect=[0, 0.1, 1, 1])
note_x = _note_center_x(fig, ax)
if whole_ad > pretrain_ad:
    diff = whole_ad - pretrain_ad
    fig.text(note_x, note_y, f"Note: The {whole_ad} AD patients in 'Whole DELCODE' include {diff} converters who had post-conversion AD scans.\nThe Pretrain pool classifies these exclusively as 'converter', leaving {pretrain_ad} pure AD patients.", ha='center', fontsize=9, bbox=dict(facecolor='lightyellow', alpha=0.5))
    note_y -= 0.09
fig.text(note_x, note_y, SCAN_ON_DISK_NOTE, ha='center', fontsize=8, style='italic', color='gray')

plt.show()

# --- Plot 1b: scans per diagnosis, grouped bar across cohorts (scan-level counterpart of Plot 1) ---
scan_cols = [c for c in diagnosis_table.columns if c.endswith('(scans)')]
plot_df_scans = diagnosis_table[scan_cols].sort_index()

FIG_WIDTH = 10  
fig, ax = plt.subplots(figsize=(FIG_WIDTH, 4.5))
plot_df_scans.plot(kind='bar', ax=ax)
for container in ax.containers:
    ax.bar_label(container, fontsize=8)
ax.set_title('Scans per diagnosis, by cohort')
ax.set_xlabel('diagnosis')
ax.set_ylabel('# scans')
ax.legend(title=None, fontsize='small')

whole_ad_scans = diagnosis_table['Whole DELCODE (scans)'].get('ad', 0)
pretrain_ad_scans = diagnosis_table['Pretrain pool (scans)'].get('ad', 0)
fig.tight_layout(rect=[0, 0.1, 1, 1])
notes = []
if whole_ad_scans > pretrain_ad_scans:
    diff_scans = whole_ad_scans - pretrain_ad_scans
    notes.append(
        f"Note: The {whole_ad_scans} AD scans in 'Whole DELCODE' include {diff_scans} post-conversion scans "
        f"belonging to converters.\nThe Pretrain pool attributes a converter's scans (pre- and post-conversion) "
        f"entirely to 'converter', leaving {pretrain_ad_scans} scans under pure AD patients."
    )


# --- Dynamic note: why converter/mci scan counts also shift Pretrain -> Downstream ---
def _scan_diagnosis_breakdown(diag):
    literal = int(whole_stats['diagnosis_scans'].get(diag, 0))
    pretrain_sum = int(pretrain_stats['pool'].groupby('diagnosis')['n_scans'].sum().get(diag, 0))
    downstream_sum = (
        int(downstream_stats['pool'].groupby('diagnosis')['n_scans'].sum().get(diag, 0))
        if downstream_stats is not None else None
    )
    patient_diag = cohorts_df.groupby('Pseudonym')['diagnosis'].apply(set)
    if diag == 'converter':
        group_pids = patient_diag[patient_diag.apply(lambda d: 'converter' in d)].index
    elif diag == 'mci':
        group_pids = patient_diag[patient_diag.apply(lambda d: 'mci' in d and 'converter' not in d)].index
    else:
        group_pids = patient_diag[patient_diag.apply(lambda d: diag in d)].index
    group_rows = cohorts_df[cohorts_df['Pseudonym'].isin(group_pids)]
    ad_labeled_rows = int((group_rows['diagnosis'] == 'ad').sum())
    on_disk_total = sum(
        len(list(MATRICES_DIR.glob(f'sub-{pid}_*_z_transformed.npz'))) for pid in group_pids
    )
    unmatched_on_disk = on_disk_total - len(group_rows)
    return {
        'literal': literal, 'pretrain_sum': pretrain_sum, 'downstream_sum': downstream_sum,
        'ad_labeled_rows': ad_labeled_rows, 'unmatched_on_disk': unmatched_on_disk,
    }


def _delta_reason(stats):
    reasons = []
    if stats['ad_labeled_rows']:
        reasons.append(f"{stats['ad_labeled_rows']} post-conversion ad-labeled visits")
    if stats['unmatched_on_disk']:
        reasons.append(f"{stats['unmatched_on_disk']} on-disk scans with no cohorts.csv row")
    return ' + '.join(reasons) if reasons else 'edge-case months'


conv_b = _scan_diagnosis_breakdown('converter')
mci_b = _scan_diagnosis_breakdown('mci')
if conv_b['downstream_sum'] is not None:
    conv_reason = _delta_reason(conv_b)
    mci_reason = _delta_reason(mci_b)
    notes.append(
        f"Note: converter scans {conv_b['literal']}→{conv_b['pretrain_sum']}→{conv_b['downstream_sum']} and "
        f"mci scans {mci_b['literal']}→{mci_b['pretrain_sum']}→{mci_b['downstream_sum']} "
        "(Whole DELCODE→Pretrain→Downstream) diverge because Pretrain counts every on-disk scan per "
        "patient regardless of per-visit label (folding in pre/post-conversion visits under one group label), "
        f"while Downstream's `_allowed_months` additionally excludes {conv_reason} (converter) and {mci_reason} (mci)."
    )

notes.append(SCAN_ON_DISK_NOTE)

combined_note = "\n\n".join(notes)
n_lines = combined_note.count('\n') + 1
rect_bottom = 0.05 + 0.025 * n_lines
fig.set_size_inches(FIG_WIDTH, 4.5 + 0.22 * n_lines)   # widen + grow height for notes
fig.tight_layout(rect=[0, rect_bottom, 1, 1])

fig.text(
    0.5, rect_bottom * 0.5,          # exact center, no note_x offset
    combined_note,
    ha='center', va='top', fontsize=8, wrap=True,
    bbox=dict(facecolor='lightyellow', alpha=0.5),
)

fig.savefig(RUN_DIR, dpi=150, bbox_inches='tight', pad_inches=0.3)
plt.show()

# --- Plot 2: scans-per-patient distribution, one panel per cohort, stacked by diagnosis ---
# Whole DELCODE has no single exclusive diagnosis per patient (cohorts.csv is visit-level), so
# reuse the Pretrain pool's exclusive per-patient label for a consistent breakdown across panels;
# patients absent from the Pretrain pool (e.g. SCD-only/relatives) are bucketed as 'other/excluded'.
patient_diag_map = (
    pretrain_stats['pool'].assign(Pseudonym=lambda d: d['Pseudonym'].astype(str))
    .set_index('Pseudonym')['diagnosis']
)
whole_diag_df = whole_stats['scans_per_patient'].rename('n_scans').to_frame()
whole_diag_df['diagnosis'] = whole_diag_df.index.map(patient_diag_map).fillna('other/excluded')
whole_crosstab = whole_diag_df.groupby(['n_scans', 'diagnosis']).size().unstack(fill_value=0)

cohort_crosstabs = [
    ('Whole DELCODE', whole_crosstab),
    ('Pretrain pool', pretrain_stats['pool'].groupby(['n_scans', 'diagnosis']).size().unstack(fill_value=0)),
]
if downstream_stats is not None:
    cohort_crosstabs.append(
        ('Downstream pool', downstream_stats['pool'].groupby(['n_scans', 'diagnosis']).size().unstack(fill_value=0))
    )

fig, axes = plt.subplots(1, len(cohort_crosstabs), figsize=(5 * len(cohort_crosstabs), 4), sharey=False)
if len(cohort_crosstabs) == 1:
    axes = [axes]
for ax, (title, crosstab) in zip(axes, cohort_crosstabs):
    _plot_diag_stack(ax, crosstab, title, '# scans per patient')
fig.suptitle('Scans-per-patient distribution, by diagnosis')
present_cols = sorted({c for _, ct in cohort_crosstabs for c in ct.columns}, key=DIAG_ORDER.index)
handles = [plt.Rectangle((0, 0), 1, 1, color=plt.cm.tab10(DIAG_ORDER.index(c))) for c in present_cols]
fig.legend(handles, present_cols, title='diagnosis', loc='upper center', bbox_to_anchor=(0.5, 1.15), ncol=len(present_cols), fontsize='small')
fig.tight_layout(rect=[0, 0.05, 1, 1])
note_x = _note_center_x(fig, axes)
fig.text(note_x, -0.03, SCAN_ON_DISK_NOTE, ha='center', fontsize=8, style='italic', color='gray')
plt.show()

# --- Plot 3: per-split diagnosis composition, pretrain vs downstream, with a labeled total ---
split_diag_tables = [('Pretrain', pretrain_stats['diagnosis_per_split'])]
if downstream_stats is not None:
    split_diag_tables.append(('Downstream', downstream_stats['diagnosis_per_split']))

fig, axes = plt.subplots(1, len(split_diag_tables), figsize=(5 * len(split_diag_tables), 4), sharey=False)
if len(split_diag_tables) == 1:
    axes = [axes]
for ax, (title, diag_per_split) in zip(axes, split_diag_tables):
    _plot_diag_stack(ax, diag_per_split, f'{title} split pool', 'split')
present_cols = sorted({c for _, t in split_diag_tables for c in t.columns}, key=DIAG_ORDER.index)
handles = [plt.Rectangle((0, 0), 1, 1, color=plt.cm.tab10(DIAG_ORDER.index(c))) for c in present_cols]
fig.legend(handles, present_cols, title='diagnosis', loc='upper center', bbox_to_anchor=(0.52, 1.1), ncol=len(present_cols), fontsize='small')
fig.tight_layout(rect=[0, 0.05, 1, 1])
note_x = _note_center_x(fig, axes)
fig.text(note_x, -0.03, SCAN_ON_DISK_NOTE, ha='center', fontsize=8, style='italic', color='gray')
plt.show()
```

### G. Why the pretrain pool's "Pretrain pool" panel above is dominated by `n_scans = 1`

`n_scans` in the pretrain split CSVs is computed from the actual `.npz` matrix files on disk
(`_count_scans` in `DATA/src/splitting/create_pretrain_data_splits.py`), not from the number
of clinical visit rows in `cohorts.csv`. The two diverge by design:

- **`ad` / `healthy`** — GAAE pretraining treats these as **static / cross-sectional**: only the
  baseline scan (`ses-01_M0`) was ever turned into a connectivity matrix, even though `cohorts.csv`
  may list several later clinical visits for the same patient. So `n_scans = 1` for every single
  `ad`/`healthy` patient — that's not a stale or wrong number, there is genuinely only one matrix
  file for them.
- **`mci` / `converter`** — these are the longitudinal/conversion-relevant classes, so every
  follow-up scan was processed into its own matrix. `n_scans` for these patients reflects the real
  multi-visit trajectory (mean ≈ 2.8 for `mci`, ≈ 3.4 for `converter`).

So in the "Pretrain pool" panel of plot E2, the `n_scans = 1` bar is exactly
`n(ad) + n(healthy) + n(mci with 1 scan) + n(converter with 1 scan)` — not a measurement error.
The table below breaks that bar down by diagnosis.

```python
# Pretrain pool: patients per (diagnosis, n_scans) — explains why n_scans=1 dominates.
pretrain_breakdown = (
    pretrain_stats['pool']
    .groupby(['diagnosis', 'n_scans'])
    .size()
    .unstack(fill_value=0)
    .reindex(['ad', 'healthy', 'mci', 'converter'])
)
pretrain_breakdown.loc['TOTAL'] = pretrain_breakdown.sum()
print('Pretrain pool — patients per (diagnosis, n_scans):')
display(pretrain_breakdown)

fig, ax = plt.subplots(figsize=(8, 4.5))
pretrain_breakdown.drop(index='TOTAL').plot(kind='bar', stacked=True, ax=ax)
ax.set_title('Pretrain pool — n_scans composition by diagnosis')
ax.set_xlabel('diagnosis')
ax.set_ylabel('# patients')
ax.legend(title='n_scans', fontsize='small', ncol=2)

# Longitudinal-data note: healthy/ad are cross-sectional (n_scans=1 for every patient),
# mci/converter are the only diagnoses with any patient at n_scans > 1.
scan_counts = [c for c in pretrain_breakdown.columns if c != 1]
cross_sectional = [
    d for d in ['healthy', 'ad']
    if d in pretrain_breakdown.index and pretrain_breakdown.loc[d, scan_counts].sum() == 0
]
longitudinal = [
    d for d in ['mci', 'converter']
    if d in pretrain_breakdown.index and pretrain_breakdown.loc[d, scan_counts].sum() > 0
]
note_y = -0.05
fig.tight_layout(rect=[0, 0.1, 1, 1])
note_x = _note_center_x(fig, ax)
if cross_sectional and longitudinal:
    fig.text(
        note_x, note_y,
        f"Note: Only {' and '.join(longitudinal)} have longitudinal data (n_scans > 1 for at least "
        f"some patients).\n{' and '.join(cross_sectional)} are cross-sectional here — every patient "
        "has exactly 1 scan.",
        ha='center', fontsize=9, bbox=dict(facecolor='lightyellow', alpha=0.5),
    )
    note_y -= 0.09
fig.text(note_x, note_y, SCAN_ON_DISK_NOTE, ha='center', fontsize=8, style='italic', color='gray')

plt.show()
```

### H. Why `converter` averages *more* scans than stable `mci`

Counter-intuitive at first: you might expect converters to have fewer usable scans since they
"leave" the MCI state partway through. They don't — `n_scans` counts every matrix file for the
patient regardless of which diagnosis label a given visit carries, and conversion does not
truncate that count.

1. **`converter` is a patient-level label, fixed once** — assigned because this patient was
   observed transitioning MCI → AD *at some point* in the study. Visits after the transition are
   relabeled `ad` in `cohorts.csv`, but their matrices still belong to this `Pseudonym` and are
   still counted.
2. **Detecting a conversion requires surviving in the study long enough to be caught** — at
   minimum a baseline MCI visit plus a later AD visit, and DELCODE's visit interval is ~12
   months, so converters accumulate several visits before *and* after the transition is recorded.
3. **`mci` ("stable") is a mixed bag** — it includes patients genuinely stable across years of
   follow-up *and* patients who simply left the study after 1–2 visits before any conversion had
   a chance to be observed. That second group pulls the `mci` average down; there's no equivalent
   drag on `converter`, because by construction a converter already proved they stuck around.

The check below confirms point 1 directly: count, per converter, how many of their scans were
taken *after* the visit where `diagnosis` first flips to `ad`.

```python
if downstream_stats is not None:
    def _visits_after_first_ad(pid):
        """How many of this patient's cohorts.csv rows come at/after their first 'ad'-labeled visit.

        Sorts by visdate (parsed date), not the 'visit' string — 'M108' < 'M12' lexicographically,
        which would silently misorder any patient followed past 100 months.
        """
        sub = cohorts_df[cohorts_df['Pseudonym'] == pid].copy()
        sub['visdate'] = pd.to_datetime(sub['visdate'], dayfirst=True)
        sub = sub.sort_values('visdate')
        diags = sub['diagnosis'].tolist()
        if 'ad' not in diags:
            return None
        first_ad_idx = diags.index('ad')
        return len(diags) - first_ad_idx

    converter_ids = downstream_stats['pool'].loc[
        downstream_stats['pool']['diagnosis'] == 'converter', 'Pseudonym'
    ].tolist()
    post_conversion = {pid: _visits_after_first_ad(pid) for pid in converter_ids}
    observed = {pid: n for pid, n in post_conversion.items() if n is not None}

    print(f'Converters with >=1 visit recorded at/after the AD relabel: '
          f'{len(observed)} / {len(converter_ids)}')
    print(f'Average # such post-conversion visits (among those): '
          f'{sum(observed.values()) / len(observed):.2f}')
    print('\n(These post-conversion visits still contribute to n_scans for that Pseudonym —')
    print(' conversion does not truncate the count, it only relabels later rows as "ad".)')

    # Compare n_scans=1 prevalence: stable mci vs converter (the "early dropout" signature).
    mci_n1 = (downstream_stats['pool'].query("diagnosis == 'mci'")['n_scans'] == 1).mean()
    conv_n1 = (downstream_stats['pool'].query("diagnosis == 'converter'")['n_scans'] == 1).mean()
    print(f"\nShare of patients with only 1 scan — mci: {mci_n1:.1%}, converter: {conv_n1:.1%}")
else:
    print('Downstream splits not present at expected paths — skipping')
```
